```python
cw_it_de_reduced_gmerged = pd.merge(
    left=cw_it_de_reduced,
    right=greenness_jrc,
    left_on="Classification_2_ID_it",
    right_on="isco08_jrc",
    how="left",
    validate="m:1",
)

# fill nans in greenness col with zeros
cw_it_de_reduced_gmerged["greenness_jrc"] = cw_it_de_reduced_gmerged[
    "greenness_jrc"
].fillna(0)

cw_it_de_reduced_gmerged
```
